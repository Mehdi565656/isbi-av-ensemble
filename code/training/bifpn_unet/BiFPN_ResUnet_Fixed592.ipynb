# BiFPN-ResUNet -- fixed 592x592

Trains a **BiFPN-ResUNet**: a BiFPN-UNet where the standard convolutional encoder
(`ConvBlock`) is replaced by a **residual encoder** (`ResidualBlock` with a 1x1 projection
shortcut). Everything else -- BiFPN neck (bifpn_ch=64), decoder (UpBlock), loss, training
protocol -- is identical to `BiFPN_Unet_64to1024_Fixed592.ipynb`.

## Architecture

- **Encoder**: `ResEncoder(base_ch=64, depth=5)` -> residual blocks at channels
  [64, 128, 256, 512, 1024]. Each block: Conv-BN-ReLU-Conv-BN + 1x1 projection shortcut
  (when in_ch != out_ch), final ReLU.
- **Neck**: BiFPN, bifpn_ch=64 (bounded -- all pyramid levels projected to 64ch before
  fusion, regardless of encoder depth, to avoid full-resolution high-channel OOM).
- **Decoder**: UpBlock (transpose-conv up + skip concat + ConvBlock) x4.
- **Head**: 1x1 conv -> 3 classes.

## Training conditions (identical to BiFPN-UNet 64-1024ch)

- AdamW (lr=5e-4, wd=2e-4) + CosineAnnealingLR (T_max=300, eta_min=1e-5)
- CEDiceLoss (0.7 CE + 0.3 SoftDice on vessel classes)
- Batch size 4, up to 300 epochs, PATIENCE_EVALS=20 (100-epoch patience window)
- Augmentation: h-flip, v-flip, rotation +/-10 deg

## Dataset

```python
class ImageDataset(data.Dataset):
    def __init__(self, root, new_size=592, supervised=True, augment=False):
        self.root = Path(root)
        self.new_size = new_size
        self.supervised = supervised
        self.augment = augment

        img_dir = self.root / 'images'
        ann_dir = self.root / 'annotation'
        self.image_paths = sorted(img_dir.glob('*.png'))
        self.label_paths = [ann_dir / f'{p.stem}.png' for p in self.image_paths] if supervised else []

        print(f"Dataset root        : {root}")
        print(f"  Images found      : {len(self.image_paths)}")
        if supervised:
            print(f"  Annotations found : {len(self.label_paths)}")
        print(f"  Augmentation      : {self.augment}")
        if self.image_paths:
            print(f"  First 5 images    : {[p.name for p in self.image_paths[:5]]}")

    @staticmethod
    def rgb_to_class_mask(rgb):
        r, g, b = rgb[:, :, 0], rgb[:, :, 1], rgb[:, :, 2]
        mask = np.zeros(rgb.shape[:2], dtype=np.int64)
        mask[(r > 180) & (g < 80) & (b < 80)]  = 1
        mask[(r < 80)  & (g < 80) & (b > 180)] = 2
        return mask

    def __len__(self): return len(self.image_paths)

    def __getitem__(self, index):
        s = self.new_size
        image = Image.open(self.image_paths[index]).convert('RGB').resize((s, s), Image.BILINEAR)
        if not self.supervised:
            return T.ToTensor()(image)
        ann = Image.open(self.label_paths[index]).convert('RGB').resize((s, s), Image.NEAREST)
        if self.augment:
            if random.random() < 0.5:
                image = image.transpose(Image.FLIP_LEFT_RIGHT)
                ann   = ann.transpose(Image.FLIP_LEFT_RIGHT)
            if random.random() < 0.5:
                image = image.transpose(Image.FLIP_TOP_BOTTOM)
                ann   = ann.transpose(Image.FLIP_TOP_BOTTOM)
            angle = random.uniform(-10, 10)
            image = image.rotate(angle, resample=Image.BILINEAR, fillcolor=(0, 0, 0))
            ann   = ann.rotate(angle,   resample=Image.NEAREST,  fillcolor=(0, 0, 0))
        return T.ToTensor()(image), torch.from_numpy(self.rgb_to_class_mask(np.array(ann)))
```

## BiFPN-ResUNet Architecture

The encoder swaps `ConvBlock` (two plain conv-BN-ReLU layers) for `ResidualBlock`
(same two layers plus a 1x1 projection shortcut). The BiFPN neck and decoder are
unchanged from `BiFPN_Unet_64to1024_Fixed592.ipynb`.

```python
# ── Residual encoder block ────────────────────────────────────────────────────

class ResidualBlock(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.block = nn.Sequential(
            nn.Conv2d(in_ch,  out_ch, 3, padding=1, bias=False),
            nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
            nn.Conv2d(out_ch, out_ch, 3, padding=1, bias=False),
            nn.BatchNorm2d(out_ch),
        )
        self.shortcut = (
            nn.Sequential(nn.Conv2d(in_ch, out_ch, 1, bias=False), nn.BatchNorm2d(out_ch))
            if in_ch != out_ch else nn.Identity()
        )
        self.relu = nn.ReLU(inplace=True)

    def forward(self, x):
        return self.relu(self.block(x) + self.shortcut(x))


class ResEncoder(nn.Module):
    """Residual encoder: one ResidualBlock per level + MaxPool downsampling."""
    def __init__(self, in_ch=3, base_ch=64, depth=5):
        super().__init__()
        chs = [base_ch * (2 ** i) for i in range(depth)]
        self.blocks = nn.ModuleList()
        self.pools  = nn.ModuleList()
        c = in_ch
        for ch in chs:
            self.blocks.append(ResidualBlock(c, ch))
            self.pools.append(nn.MaxPool2d(2))
            c = ch
        self.out_channels = chs

    def forward(self, x):
        feats = []
        for block, pool in zip(self.blocks, self.pools):
            x = block(x)
            feats.append(x)
            x = pool(x)
        return feats


# ── Plain conv block (used in BiFPN fusion convs and decoder) ─────────────────

class ConvBlock(nn.Module):
    def __init__(self, in_ch, out_ch):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv2d(in_ch, out_ch, 3, padding=1, bias=False),
            nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
            nn.Conv2d(out_ch, out_ch, 3, padding=1, bias=False),
            nn.BatchNorm2d(out_ch), nn.ReLU(inplace=True),
        )
    def forward(self, x): return self.conv(x)


# ── BiFPN neck (unchanged from BiFPN_Unet_64to1024_Fixed592.ipynb) ────────────

class BiFPN(nn.Module):
    def __init__(self, channels, num_layers=1, eps=1e-4):
        super().__init__()
        self.eps = eps
        self.num_levels = len(channels)
        self.num_layers = num_layers
        self.convs_td = nn.ModuleList([ConvBlock(c, c) for c in channels])
        self.convs_bu = nn.ModuleList([ConvBlock(c, c) for c in channels])
        self.w_td = nn.Parameter(torch.ones(num_layers, self.num_levels, 2))
        self.w_bu = nn.Parameter(torch.ones(num_layers, self.num_levels, 3))

    def forward(self, feats):
        P = feats
        for l in range(self.num_layers):
            td = [None] * self.num_levels
            td[-1] = P[-1]
            for i in reversed(range(self.num_levels - 1)):
                w  = F.relu(self.w_td[l, i]); w = w / (w.sum() + self.eps)
                up = F.interpolate(td[i+1], size=P[i].shape[-2:], mode='bilinear', align_corners=False)
                td[i] = self.convs_td[i](w[0]*P[i] + w[1]*up)
            bu = [None] * self.num_levels
            bu[0] = td[0]
            for i in range(1, self.num_levels):
                w    = F.relu(self.w_bu[l, i]); w = w / (w.sum() + self.eps)
                down = F.max_pool2d(bu[i-1], 2)
                if down.shape[-2:] != td[i].shape[-2:]:
                    down = F.interpolate(down, size=td[i].shape[-2:], mode='bilinear', align_corners=False)
                bu[i] = self.convs_bu[i](w[0]*P[i] + w[1]*td[i] + w[2]*down)
            P = bu
        return P


# ── Decoder block (unchanged) ─────────────────────────────────────────────────

class UpBlock(nn.Module):
    def __init__(self, in_ch, skip_ch, out_ch):
        super().__init__()
        self.up   = nn.ConvTranspose2d(in_ch, out_ch, 2, stride=2)
        self.conv = ConvBlock(out_ch + skip_ch, out_ch)

    def forward(self, x, skip):
        x = self.up(x)
        if x.shape[-2:] != skip.shape[-2:]:
            x = F.interpolate(x, size=skip.shape[-2:], mode='bilinear', align_corners=False)
        return self.conv(torch.cat([skip, x], dim=1))


# ── Full model ────────────────────────────────────────────────────────────────

class BiFPNResUNet(nn.Module):
    def __init__(self, in_channels=3, num_classes=3,
                 base_ch=64, depth=5, bifpn_ch=64, num_bifpn_layers=1):
        super().__init__()
        self.encoder    = ResEncoder(in_ch=in_channels, base_ch=base_ch, depth=depth)
        self.proj       = nn.ModuleList([nn.Conv2d(c, bifpn_ch, 1) for c in self.encoder.out_channels])
        self.bifpn      = BiFPN([bifpn_ch] * depth, num_layers=num_bifpn_layers)
        self.center     = ConvBlock(bifpn_ch, bifpn_ch)
        self.up_blocks  = nn.ModuleList([UpBlock(bifpn_ch, bifpn_ch, bifpn_ch) for _ in range(depth-1)])
        self.final_conv = nn.Conv2d(bifpn_ch, num_classes, 1)

    def forward(self, x):
        enc = self.encoder(x)
        enc = [p(f) for p, f in zip(self.proj, enc)]
        fused = self.bifpn(enc)
        x = self.center(fused[-1])
        for i, up in enumerate(self.up_blocks):
            x = up(x, fused[-2-i])
        return self.final_conv(x)
```

## Loss: Weighted CE + Soft Dice

```python
class SoftDiceLoss(nn.Module):
    def __init__(self, classes=VESSEL_CLASSES, eps=1e-6):
        super().__init__()
        self.classes = classes; self.eps = eps

    def forward(self, logits, labels):
        probs = torch.softmax(logits, dim=1)
        losses = []
        for c in self.classes:
            p = probs[:, c]; t = (labels == c).float()
            tp = (p * t).sum(dim=(1, 2))
            fp = (p * (1 - t)).sum(dim=(1, 2))
            fn = ((1 - p) * t).sum(dim=(1, 2))
            losses.append((1 - (2*tp + self.eps) / (2*tp + fp + fn + self.eps)).mean())
        return torch.stack(losses).mean()


class CEDiceLoss(nn.Module):
    def __init__(self, class_weights=None, dice_weight=0.3):
        super().__init__()
        self.ce = nn.CrossEntropyLoss(weight=class_weights)
        self.dice = SoftDiceLoss()
        self.dice_weight = dice_weight

    def forward(self, logits, labels):
        return (1 - self.dice_weight) * self.ce(logits, labels.long()) + \
               self.dice_weight * self.dice(logits, labels.long())
```

## Metrics

```python
def dice_coef_per_class(preds, labels, num_classes=NUM_CLASSES, eps=1e-6):
    per_class = {}
    for c in range(num_classes):
        p = (preds == c).float(); l = (labels == c).float()
        tp = float((p*l).sum()); fp = float((p*(1-l)).sum()); fn = float(((1-p)*l).sum())
        per_class[c] = 2*tp / (2*tp + fp + fn + eps)
    return per_class, float(np.mean([per_class[c] for c in VESSEL_CLASSES]))
```

## Training loop

```python
def test_model(model, dataloader, verbose=True):
    model.eval()
    total_dice, n = 0.0, 0
    with torch.no_grad():
        for inputs, labels in dataloader:
            inputs, labels = inputs.to(device), labels.to(device)
            preds = torch.argmax(torch.softmax(model(inputs), dim=1), dim=1)
            _, vessel_dice = dice_coef_per_class(preds, labels)
            total_dice += vessel_dice; n += 1
    final_dice = total_dice / max(n, 1)
    if verbose:
        print(f'Mean vessel Dice on test set: {final_dice:.4f}')
    return final_dice


def train_model(model, criterion, optimizer, scheduler, train_loader, test_loader,
                num_epochs, eval_every=5,
                checkpoint_path='models/best_model_bifpn_resunet_592.pth',
                history_path='training_history_bifpn_resunet_592.csv',
                patience_evals=20):
    best_dice, best_state = -1.0, copy.deepcopy(model.state_dict())
    epoch_losses, eval_history = [], []
    evals_since_best = 0

    os.makedirs(os.path.dirname(checkpoint_path) or '.', exist_ok=True)
    with open(history_path, 'w') as f:
        f.write('epoch,test_vessel_dice,is_best\n')

    for epoch in range(num_epochs):
        model.train()
        batch_losses = []
        for inputs, labels in train_loader:
            inputs, labels = inputs.to(device), labels.to(device)
            optimizer.zero_grad()
            loss = criterion(model(inputs), labels)
            loss.backward(); optimizer.step()
            batch_losses.append(loss.item())

        scheduler.step()
        mean_loss = sum(batch_losses) / len(batch_losses)
        epoch_losses.append(mean_loss)

        is_last = (epoch == num_epochs - 1)
        if (epoch + 1) % eval_every == 0 or is_last:
            dice = test_model(model, test_loader, verbose=False)
            eval_history.append((epoch + 1, dice))
            is_best = dice > best_dice
            if is_best:
                best_dice = dice
                best_state = copy.deepcopy(model.state_dict())
                torch.save(best_state, checkpoint_path)
                evals_since_best = 0
            else:
                evals_since_best += 1
            with open(history_path, 'a') as f:
                f.write(f'{epoch+1},{dice:.6f},{int(is_best)}\n')
            print(f'Epoch {epoch+1}/{num_epochs}  Loss: {mean_loss:.4f}  '
                  f'Test vessel Dice: {dice:.4f}  (best so far: {best_dice:.4f})')
            if patience_evals is not None and evals_since_best >= patience_evals:
                print(f'Early stopping: no improvement in {patience_evals} eval checks '
                      f'({patience_evals * eval_every} epochs).')
                break
        else:
            print(f'Epoch {epoch+1}/{num_epochs}  Loss: {mean_loss:.4f}')

    print('Training complete.')
    model.load_state_dict(best_state)
    return model, epoch_losses, eval_history, best_dice
```

## Memory profile

```python
profile_model = BiFPNResUNet(in_channels=3, num_classes=NUM_CLASSES,
                              base_ch=64, depth=5, bifpn_ch=64).to(device)
n_params = sum(p.numel() for p in profile_model.parameters())
print(f"Model parameters: {n_params:,}")

torch.cuda.reset_peak_memory_stats(device)
profile_model.train()
dummy_in  = torch.randn(4, 3, 592, 592, device=device)
dummy_lbl = torch.randint(0, NUM_CLASSES, (4, 592, 592), device=device)
dummy_out = profile_model(dummy_in)
CEDiceLoss()(dummy_out, dummy_lbl).backward()
peak_mb = torch.cuda.max_memory_allocated(device) / 1024**2
total_mb = torch.cuda.get_device_properties(device).total_memory / 1024**2
print(f"Peak GPU memory (batch=4, 592x592): {peak_mb:.0f} MB / {total_mb:.0f} MB total")

del profile_model, dummy_in, dummy_lbl, dummy_out
gc.collect(); torch.cuda.empty_cache()
```

## Training (300 epochs)

```python
EPOCHS         = 300
BATCH_SIZE     = 4
SEED           = 42
EVAL_EVERY     = 5
DICE_WEIGHT    = 0.3
PATIENCE_EVALS = 20

full_dataloader = data.DataLoader(
    train_dataset_full, shuffle=True, batch_size=BATCH_SIZE,
    num_workers=4, pin_memory=True, persistent_workers=True)

print(f'\n=== Training BiFPN-ResUNet on {len(train_dataset_full)} images ===')

torch.manual_seed(SEED); random.seed(SEED); np.random.seed(SEED)

# Class weights
pixel_counts = Counter()
plain_ds = ImageDataset(root=train_root, supervised=True, augment=False)
for _, lbl in plain_ds:
    for c in range(NUM_CLASSES):
        pixel_counts[c] += int((lbl == c).sum())
total    = sum(pixel_counts.values())
freqs    = torch.tensor([pixel_counts[c] / total for c in range(NUM_CLASSES)])
inv_freq = 1.0 / (freqs + 1e-6)
class_weights = (inv_freq / inv_freq[1:].mean()).to(device)
print("Class pixel %:", {CLASS_NAMES[c]: f"{100*freqs[c].item():.2f}%" for c in range(NUM_CLASSES)})
print("Class weights:", class_weights)

model     = BiFPNResUNet(in_channels=3, num_classes=NUM_CLASSES,
                         base_ch=64, depth=5, bifpn_ch=64, num_bifpn_layers=1).to(device)
criterion = CEDiceLoss(class_weights, dice_weight=DICE_WEIGHT)
optimizer = optim.AdamW(model.parameters(), lr=5e-4, weight_decay=2e-4)
scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=EPOCHS, eta_min=1e-5)

n_params = sum(p.numel() for p in model.parameters())
print(f"Model parameters: {n_params:,}")

save_path = 'models/best_model_bifpn_resunet_592.pth'
model, training_losses, eval_history, best_dice = train_model(
    model, criterion, optimizer, scheduler,
    full_dataloader, test_loader,
    num_epochs=EPOCHS, eval_every=EVAL_EVERY,
    checkpoint_path=save_path,
    history_path='training_history_bifpn_resunet_592.csv',
    patience_evals=PATIENCE_EVALS)

final_dice = test_model(model, test_loader)
print(f"\nBest test vessel Dice during training        : {best_dice:.4f}")
print(f"Final-epoch test vessel Dice (no selection)   : {eval_history[-1][1]:.4f}")
print(f"ResUNet (300 epochs)                          : 0.8103")
print(f"BiFPN-UNet (64-1024ch, 300 epochs)            : 0.8092")
print(f"Standard U-Net baseline                       : 0.8023")
print(f"{'Beats' if best_dice > 0.8023 else 'Does not beat'} the Standard U-Net baseline.")
print(f"Model saved to {save_path}")

fig, ax1 = plt.subplots(figsize=(8, 6))
ax1.plot(range(1, len(training_losses)+1), training_losses, color='tab:blue', label='Train Loss')
ax1.set_xlabel('Epoch'); ax1.set_ylabel('Train Loss', color='tab:blue')
ax1.tick_params(axis='y', labelcolor='tab:blue')
ax2 = ax1.twinx()
eval_epochs, eval_dices = zip(*eval_history)
ax2.plot(eval_epochs, eval_dices, color='tab:orange', marker='o', label='Test Vessel Dice')
ax2.axhline(0.8023, color='gray', linestyle='--', label='U-Net baseline (0.8023)')
ax2.axhline(0.8103, color='green', linestyle='--', label='ResUNet (0.8103)')
ax2.set_ylabel('Test Vessel Dice', color='tab:orange')
ax2.tick_params(axis='y', labelcolor='tab:orange')
fig.legend(loc='lower center', bbox_to_anchor=(0.5, -0.05), ncol=4)
plt.title(f'BiFPN-ResUNet (fixed 592x592) | Best Dice: {best_dice:.4f}')
fig.tight_layout()
plt.savefig('training_diagnostics_bifpn_resunet_592.png', dpi=150, bbox_inches='tight')
plt.show()
```
